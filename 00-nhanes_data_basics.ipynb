```python
import pandas as pd

# Set base path
base_path = "data/"

# Load each file
alq = pd.read_sas(base_path + "ACQ_I.XPT")
demo = pd.read_sas(base_path + "DEMO_I.XPT")

# Save them as CSV
alq.to_csv(base_path + "ALQ_I.csv", index=False)
demo.to_csv(base_path + "DEMO_I.csv", index=False)

print("Files successfully converted to CSV.")
```

```python
# Merge on SEQN
merged = demo.merge(alq, on="SEQN")

# Save the merged dataset
merged.to_csv(base_path + "nhanes_2015_2016.csv", index=False)

print("Merged file saved as nhanes_2015_2016.csv")
```
