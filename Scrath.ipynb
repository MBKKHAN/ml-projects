```python
def fib2(n):
    if n<=1:return n
    a,b=0,1
    for i in range(1,n+1):
        n = a + b
        a = b
        b = n
    return a
```

```python
class Polygon:
    def __init__(self,no_of_sides):
        print("Creating a polygon of sides",no_of_sides)
        self.n = no_of_sides
        self.sides = [0 for i in range(no_of_sides)]     
    def set_sides(self,sides):
        self.sides= sides
    def show_sides(self):
        for i in self.sides:
            print("Sides:",i)
```

```python
p = Polygon(3)
print(p.n)
```

```python
p.set_sides([3,4,5])
```

```python
p.show_sides()
```

```python
class Triangle(Polygon):
    def __init__(self):
        Polygon.__init__(self,3)
    def get_area(self):
        a,b,c = self.sides
        s = (a + b + c)/3
        area = (s*(s-a)*(s-b)*(s-c))**0.5
        return area
```

```python
sides = [10,14,13]
t = Triangle()
```

```python
t.set_sides(sides)
t.show_sides()
```

```python
t.get_area()
```
